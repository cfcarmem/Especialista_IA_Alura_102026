Rodar localmente IA


Nesta aula, vamos explorar uma abordagem diferente para lidar com Inteligências Artificiais (IAs) em Python, executando-as localmente. Pode-se questionar: por que fazer isso? Existem diversos motivos. Muitas vezes, lidamos com dados privados e sensíveis, e mesmo confiando em empresas que oferecem esses serviços, como Google, OpenAI e Anthropic, pode haver razões pessoais ou empresariais, ou até mesmo leis de proteção de dados, que nos levam a preferir rodar esses modelos localmente.

Outro motivo seria o desenvolvimento de um modelo próprio, do zero, ou a utilização de modelos disponíveis online, como o GPT OSS de 20 bilhões de parâmetros, que é um modelo de código aberto. Podemos realizar um fine tuning (ajuste fino) nesses modelos para especializá-los em áreas específicas, como jurídico, financeiro ou recursos humanos, por exemplo. Se o modelo for menor, podemos executá-lo localmente em nossa máquina ou em um cluster de GPUs.


1. O que é o LM Studio?
O LM Studio é um programa que permite baixar, instalar e executar modelos de inteligência artificial diretamente no seu computador, sem precisar enviar cada pergunta para serviços externos, como ChatGPT, Gemini ou Groq.

Em outras palavras, você pode ter um modelo de IA funcionando localmente na sua máquina.

Por exemplo, pode executar modelos como:
- Llama
- Mistral
- Qwen
- Gemma
- DeepSeek
- gpt-oss

Instalei o LM Studio e baixei o modelo google Gemma 3 1B Q4 . O B quer dizer um milhao, ou seja, este modelo tem 1 bilhão de parametros.
 Parâmetros são valores numéricos aprendidos durante o treinamento da inteligência artificial. Eles ajudam o modelo a interpretar perguntas e produzir respostas. (São os pesos)
 O Q4 quer dizer que é formado por 4 numeros,tipo inteiro, 9999

Quanto maior, melhor?
Não necessariamente!
- 1B: menor, geralmente mais rápido e com menor consumo de memória.
- 3B e 4B: podem oferecer respostas mais elaboradas.
- 7B e 8B: podem apresentar melhor capacidade em tarefas complexas, mas exigem mais recursos.
- 70B: modelos grandes, que exigem bastante memória e processamento.
Uma observação: a quantidade de parâmetros não determina sozinha a qualidade. A arquitetura, o treinamento e a quantização também influenciam.
