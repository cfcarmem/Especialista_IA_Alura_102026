Criamos a chave API gemini no google ai studio https://aistudio.google.com/app/api-keys e adicionamos em Secrets. Em seguid guardamos na variável de ambiente  na maquina e tem que ter o nome GOOGLE_API_KEY

```python
import os  #biblioteca para guardar a chave em variável de ambiente no python
from google import genai #importando o gemini

from google.colab import userdata
os.environ["GOOGLE_API_KEY"]=userdata.get('gemini_api')

#vamos conectar o pytho no gemini

client = genai.Client()

#exemplo enviando a pergunta para a IA gemini
resposta=client.models.generate_content(model="gemini-3.5-flash-lite",contents="O que é sistema embarcado?")
print(resposta.text)

```

CONECTANDO  AO GEMINI **CHATBOT**

```python
instancia = client.chats.create(model="gemini-3.5-flash-lite")
resposta = instancia.send_message("O que é torque? Me explique de forma sucinta.")
print(resposta.text)
```

```python
instancia.get_history()
```

**CHATBOT COM PERGUNTA E RESPOSTA**
-pegar prompt do usuario


```python
prompt = input("Digite a sua pergunta")
#vamos usar laços de repetição
while prompt != 'fim':
  resposta = instancia.send_message(prompt)
  print(resposta.text)
  print("/n")   #quebra de linha
  prompt = input("Digite a sua pergunta")
```

*   Loop numa lista





```python
lista_numeros =[1,2,4,5,6,8,10]
list.append(lista_numeros,12)
list.insert(lista_numeros,2,13)
print(lista_numeros)

#ver quantas vezes elemento aparece na lista
lista2 = list.count(lista_numeros,20)

#percorrer a lista, mas os cinco primeiros
n=0
while n <= 5:
  print(lista_numeros[n])
  if n == 2:
    lista_numeros.remove(n)
  n = n+1
print(lista_numeros)

#mostrar a lista  posiçõ 1 até n-1 (o último valor depois do : é excludente. Ao invés de 3, ele vai trazer 2)
print(lista_numeros[1:3+1])

#A partir do 1
print(lista_numeros[1:])

#até a posição 3 (3-1.  de 0 a 2)
print(lista_numeros[:3])

#list.extend mais de um elemento
lista_numeros.extend([52,53,56])
print(lista_numeros)
```
