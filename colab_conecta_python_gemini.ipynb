```python
#3 - obter respostas a partir de um LLM usando CSV para criar arquivo

#vamos usar groq. Já foi carregado acima

from groq import Groq

client2 = Groq()

def perguntado_LLM(pergunta, npergunta):
    completion = client2.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": f"""Vou lhe mandar uma pergunta. Responder de forma sucinta, no maximo 5 linhas : {pergunta}"""
            }
        ],
        temperature=1,
        max_completion_tokens=2048,
        top_p=1,
        reasoning_effort="medium",
        stream=True,
        stop=None
    )


    with open("respostas.csv", "a", newline="", encoding="utf-8-sig") as arquivo:
          escritor = csv.writer(arquivo, delimiter=";")

          resposta_completa = ""

          for chunk in completion:
              resposta = chunk.choices[0].delta.content or ""
              # print(resposta, end="")
              resposta_completa += resposta

          escritor.writerow([f"{npergunta}-{pergunta}", resposta_completa])


    print(f"\nResposta da pergunta {npergunta} gravada no TXT.")




```

```python


client2 = Groq()

def perguntado_LLM_GROQ_PANDAS(pergunta, npergunta):
    completion = client2.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": (
                    "Responda de forma sucinta, no máximo 5 linhas: "
                    + pergunta
                )
            }
        ],
        temperature=1,
        max_completion_tokens=2048,
        top_p=1,
        reasoning_effort="medium",
        stream=True
    )

    resposta_completa = ""

    for chunk in completion:
        resposta = chunk.choices[0].delta.content or ""
        resposta_completa += resposta

    # Criar DataFrame com a pergunta e a resposta
    df = pd.DataFrame([{
        "pergunta": f"{npergunta}-{pergunta}",
        "resposta": resposta_completa
    }])

    arquivo_csv = "respostas_groq_pandas.csv"

    escrever_cabecalho = (
        not os.path.exists(arquivo_csv)
        or os.path.getsize(arquivo_csv) == 0
    )

    # Gravar usando Pandas
    df.to_csv(
        arquivo_csv,
        mode="a",
        index=False,
        header=escrever_cabecalho,
        sep=";",
        encoding="utf-8",
        lineterminator="\n"
    )

    print(f"\nResposta da pergunta {npergunta} gravada no CSV.")
```
