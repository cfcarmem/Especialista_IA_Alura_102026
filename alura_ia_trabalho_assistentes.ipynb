Planejamento e resolução de problemas

Nesta aula, aprendemos que o planejamento detalhado é uma etapa fundamental antes de efetivamente criar um assistente ou GPT. O objetivo principal é estruturar o raciocínio do assistente para que ele resolva um problema específico de forma eficiente.

O processo de planejamento abordado inclui os seguintes pontos:

Definição do problema: Identificar claramente qual tarefa o assistente deve resolver, como no exemplo dado de busca de hotéis para conferências, onde se busca equilibrar orçamento e qualidade.

Estabelecimento de critérios e formato de saída: Definir quais informações são essenciais para a busca, como orçamento, localização e disponibilidade de serviços (Wi-Fi, café da manhã), além de especificar que o formato de entrega deve ser uma tabela para facilitar a visualização e comparação.

Consideração de necessidades específicas: Incluir no planejamento requisitos de acessibilidade e necessidades particulares dos membros da equipe, garantindo que o assistente seja inclusivo e atenda às demandas reais dos usuários.

Uso da matriz CSD: Utilizar a estrutura de Certezas, Suposições e Dúvidas para organizar o comportamento do assistente:
Certezas: Dados obrigatórios que o assistente deve sempre exigir, como datas de check-in/out e orçamento.
Suposições: Aspectos importantes que podem não ser informados pelo usuário, mas que o assistente deve ser instruído a perguntar.
Dúvidas: Pontos que ainda precisam de validação com o restante da equipe antes de serem implementados nas instruções do assistente.

Essa etapa de planejamento é um processo iterativo, onde o levantamento dessas premissas ajuda a tornar o assistente mais robusto, mesmo que não garanta o acerto total na primeira tentativa.
