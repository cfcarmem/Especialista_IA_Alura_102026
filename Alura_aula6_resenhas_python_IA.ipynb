**3 - Tratamento de erros com Try Except Finally    26-10-26**

Tipos de erro:
valueError = erro de valor
typeErro = erro de tipos de dados, exemplo: tentar somar duas variáveis x = 1  e y = 'Maria'   x+y dará erro, pois são tipos diferentes para soma.
Existem inúmeras exceções em Python


```python
#O finally será executando independente de dar erro ou não. Geralmente usamos quando precisamos fechar um arquivo.

try:
    pessoas = 6
    pessoas_2 = "Mário, João, Silvia, Maria"
    pessoas + pessoas_2
except ValueError:
    print("Você não digitou um número inteiro válido")
except TypeError:
    print("Você não pode somar um número inteiro com uma string")
except Exception as e:
    print(3)
finally:
    print('Processo Finalizado')
```

```python
import pandas as pd
```

#Vamos montar a coluna Análise com valores aleátorios entre Neutro, Positivo, Negativo

```python
import random
```

```python
lista_sentimentos = ["Positiva", "Negativa", "Neutra"]
```

```python
df_arquivo["Sentimento"] = [
    random.choice(lista_sentimentos)
    for a in range(len(df_arquivo))
]
```

```python
df_arquivo
```

```python
#Vamos selecionar somente a coluna Sentinto e guardar em um df, somente Sentimentos == Negativa

df_reviews_negativas = df_arquivo[df_arquivo["Sentimento"] == "Negativa"]
```

```python
#Visualizar df_reviews_negativas
df_reviews_negativas
```

```python
#Agora vamos pegar somente a coluna com a review
df_coluna_negativa = df_reviews_negativas["reviewText"]
```

```python
#Visualizar a coluna df_coluna_negativa
df_coluna_negativa
```

```python
#Agora, vamos juntar todo o texto em uma só linha para passarmos para IA
resenhas_negativas_unidas = "#####".join(df_coluna_negativa)
```

```python
#importar gemini
import os  #biblioteca para guardar a chave em variável de ambiente no python
from google import genai #importando o gemini

from google.colab import userdata
os.environ["GOOGLE_API_KEY"]=userdata.get('gemini_api')

#vamos conectar o pytho no gemini

client = genai.Client()
```

```python
#chamando a IA para categorizar as resenhas
lista_categorizacao = []
resposta_de_categorias = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=f"""
    Você é um analista de dados.

    Vou te passar muitas resenhas negativas de análises de um produto,
    separadas por '#####', e quero que você encontre 5 categorias
    diferentes para os tipos de reclamações.

    Quero que você retorne as 5 categorias.

    Cada categoria deve ser definida por APENAS uma palavra.

    Não escreva nada além das 5 categorias.

    As categorias devem ser separadas por vírgula,
    em letras minúsculas e sem acentos gráficos.

    Exemplo:
    eletronicos, roupas, alimentos, viagens, brinquedos

    Aqui estão as resenhas negativas:
    {resenhas_negativas_unidas}
    """
)
#print(resposta_de_categorias.text)
lista_categorizacao.append(resposta_de_categorias.text)
```

```python
lista_categorizacao
```

#Aula 6 - Revisando categorização de resenhas

```python
#Vmos trabalhar com JSON

dicionario_classifica_resenhas = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=f"""
    Você é um analista de dados.

    Vou te passar muitas resenhas negativas de análises de um produto,
    separadas por '#####', e quero que você encontre 5 categorias
    diferentes para os tipos de reclamações.

    Cada categoria deve ser definida por APENAS uma palavra.

    Quero em letras minúsculas e sem acentos gráficos.
    Depois, quero que me retorne APENAS um texto no formato JSON, CONTENDO 3 chaves:
    1 - resenha_original em ingles
    2 - resenha_pt  que deve conter resenha traduzida para portugues
    3 - categoria da resenha dentre as cinco definidas


    Exemplo:
     {{
    'resenha_original': 'I didn`t like the color of the product',
    'resenha_pt': 'Eu não gostei da cor do produto',
    'categoria': 'design'
    }}

    Aqui estão as resenhas negativas:
    {resenhas_negativas_unidas}
    """
)
json_resenhas_classificadas = dicionario_classifica_resenhas.text
```
